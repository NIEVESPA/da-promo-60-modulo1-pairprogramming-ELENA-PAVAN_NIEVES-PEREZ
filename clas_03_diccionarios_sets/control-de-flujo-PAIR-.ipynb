```python
#juego animal secreto

opcion = input('dime un animal')
intento = 3
animal_secreto = 'gato'

if opcion == animal_secreto:
    print('lo has adivinado')

else:
    print('no lo has adivinado')
    while intento >0:
        opcion = input('dime un animal')
        intento -= 1
        print(f'te quedan {intento} intentos')

        if opcion == animal_secreto:
            print('lo has adivinado')
            intento = 0 #marcamos este 0 en acierto para que se detenga el bucle. sino seguiria preguntando. nosotros mismo controlamos que esa variable sea 0. entrar en un false y detener el while
        else:
            print('has perdido')
```
