```python
# Calcular el producto de los elementos de una lista

# Descripción: Dada una lista de números, utiliza un bucle for para calcular el producto de todos los elementos.
```

```python
lista = [1, 2, 3, 4]
lista_produc = []

for i in lista:
    producto = i * 2
    lista_produc.append(producto)
print(lista_produc)
```
